```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
# Directory where your training and validation images are stored
train_dir = "Pakistani_currency/train"
validation_dir = 'Pakistani_currency/val'


```

```python
# Image Data Generators for augmenting training and validation data
train_datagen = ImageDataGenerator(rescale=1./255, rotation_range=40, width_shift_range=0.2, height_shift_range=0.2, shear_range=0.2, zoom_range=0.2, horizontal_flip=True, fill_mode='nearest')
val_datagen = ImageDataGenerator(rescale=1./255)

```

```python
# Load images and apply transformations
train_generator = train_datagen.flow_from_directory(train_dir, target_size=(150, 150), batch_size=16, class_mode='categorical')
validation_generator = val_datagen.flow_from_directory(validation_dir, target_size=(150, 150), batch_size=16, class_mode='categorical')

```

```python
# Model architecture
model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(150, 150, 3)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(128, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(512, activation='relu'),
    layers.Dense(14, activation='softmax')  # Assuming 5 different types of notes
])

```

```python
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
history = model.fit(train_generator, steps_per_epoch=100, epochs=400, validation_data=validation_generator, validation_steps=50)
```

```python

# Save the model
model.save('pakistani_currency_model.h5')
```
